```python
def moving_packet(x):
        k0 = 5.0 # Initial momentum to the right
        return np.exp(-(x + 5)**2 / 2) * np.exp(1j * k0 * x)
        
def barrier(x):
        return step_potential(x, step_pos=0.0, v_left=0.0, v_right=15.0)

# Run Simulation
simulate_1D(V_func=barrier, 
                psi_0_func=moving_packet, 
                dx=0.05, 
                dt=0.01, 
                x_left=-15.0, 
                x_right=15.0, 
                frames=300, 
                methods=["split-operator"],
                output_format='video'
            )
```
